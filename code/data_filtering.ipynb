```python
import pandas as pd

BASE = "/home/ubuntu/string-matching-algorithm/"

def load(name):
    try:
        return pd.read_csv(BASE + name, encoding="utf-8-sig")
    except UnicodeDecodeError:
        return pd.read_csv(BASE + name, encoding="latin-1")

imdb = load("imb_title_set_updated.csv")
raw  = load("amrld_input_title_set (1).csv")

for name, df in [("IMDB", imdb), ("RAW INPUT", raw)]:
    print(f"\n=== {name} ===  rows, cols: {df.shape}")
    print(df.dtypes.to_string())
    print("empty values:", df.isna().sum().to_dict())
    display(df.head(10))
```

```python
# IMDb: only tvSeries rows, only the title column
imdb_tv = (
    imdb.loc[imdb["TitleType"] == "tvSeries", ["TitleUS"]]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

# Raw: only the Nielsen program name
raw_names = (
    raw[["nielsen_program_name"]]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

print("TitleType counts in the full IMDb file:")
print(imdb["TitleType"].value_counts().to_string())
print(f"\nimdb_tv:   {len(imdb_tv):,} unique tvSeries titles")
print(f"raw_names: {len(raw_names):,} unique Nielsen names (from {len(raw):,} rows)")

display(imdb_tv.sample(10, random_state=0))
display(raw_names.sample(10, random_state=0))
```

```python
imdb_tv.download = imdb_tv.to_csv(BASE + "imdb_tv.csv", index=False, encoding="utf-8-sig")
```

```python
raw_names.download = raw_names.to_csv(BASE + "raw_names.csv", index=False, encoding="utf-8-sig")
```
